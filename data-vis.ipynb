```python
import matplotlib
import mne
import pyxdf
import numpy as np

%matplotlib qt
```

```python
data = pyxdf.load_xdf("sub-Helen_ses-S001_task-Default_run-001_eeg.xdf")
```

```python
time = data[0][3]["time_stamps"]
eeg = data[0][3]["time_series"]
scale_eeg = 1e-6
n_channels = np.size(eeg, 1)
srate = float(data[0][3]["info"]["nominal_srate"][0])
info = mne.create_info(n_channels, srate, ch_types="eeg")
raw  = mne.io.RawArray(scale_eeg * eeg.T, info)
```

```python
raw.plot()
```
